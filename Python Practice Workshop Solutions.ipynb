```python
# Live refactoring: CreateUser

def normalize_username(user_name):
    return user_name.strip().lower()


def normalize_email(email):
    return email.strip().lower()


def is_valid_user(user_name, email):
    return len(user_name) >= 6 and "@" in email


def create_user(user_name, email):
    user_name = normalize_username(user_name)
    email = normalize_email(email)

    if not is_valid_user(user_name, email):
        return None

    return {
        "username": user_name,
        "email": email,
    }


user = create_user(
    "  Alex_21  ",
    " Alex@Example.COM ",
)

print(user)
```

```python
# Exercise 1: Refactor LoginUser

def normalize_username(user_name):
    return user_name.strip().lower()


def user_exists(user_name, users):
    return user_name in users


def password_matches(user_name, password, users):
    return users[user_name] == password


def login_user(user_name, password, users):
    user_name = normalize_username(user_name)

    if not user_exists(user_name, users):
        return False

    if not password_matches(user_name, password, users):
        return False

    return True


users = {
    "alex_21": "Python2026",
    "maya_dev": "SecurePass9",
}

print(login_user(" Alex_21 ", "Python2026", users))
print(login_user("alex_21", "wrong", users))
print(login_user("unknown", "test", users))
```

```python
# Exercise 2: Python built-in tools

usernames = ["alex_21", "sam99", "maya_dev", "dev_01"]
failed_attempts = [1, 3, 0, 2]

for number, user_name in enumerate(usernames, start=1):
    print(number, user_name)

print()

for user_name, attempts in zip(usernames, failed_attempts):
    print(f"{user_name}: {attempts} failed attempts")

print()

sorted_users = sorted(
    zip(usernames, failed_attempts),
    key=lambda user: user[1],
    reverse=True,
)

print(sorted_users)

has_locked_user = any(
    attempts >= 3
    for attempts in failed_attempts
)

print("Any user with 3 or more failed attempts:", has_locked_user)
```

```python
# Exercise 4: Debugging average_attempts

def average_attempts(attempts):
    numeric_attempts = [
        int(attempt)
        for attempt in attempts
    ]

    return sum(numeric_attempts) / len(numeric_attempts)


attempts = ["1", "2", "3", "2"]

average = average_attempts(attempts)
print(average)
```

```python
# Exercise 5: Debugging Merge Sorted Array (LeetCode 88)

def merge(nums1, m, nums2, n):
    i = m - 1
    j = n - 1
    k = m + n - 1

    # Fill nums1 from the end so we do not overwrite values we still need.
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1


nums1 = [4, 5, 6, 0, 0, 0]
merge(nums1, 3, [1, 2, 3], 3)
print(nums1)
```

```python
# Exercise 6: Debugging Top K Frequent Elements (LeetCode 347)

from collections import Counter


def top_k_frequent(nums, k):
    frequency = Counter(nums)
    ranked = sorted(frequency, key=lambda num: frequency[num], reverse=True)
    result = ranked[:k]
    return result


nums = [1, 1, 1, 2, 2, 3]
k = 2
print(top_k_frequent(nums, k))
```
