# 1. Cleaner and Reusable Python

## Function definitions

**Syntax**

```python
def function_name(parameters):
    statements
    return value
```

A function groups related logic under a clear name.

**PEP 8 naming:** function names normally use lowercase words separated by underscores.

**Design guideline:** keep a function focused on one clear responsibility whenever possible.

### Starting point: function to refactor

The code below intentionally uses non-PEP 8 names such as `CreateUser` and `UserName`. These are corrected during the live refactoring.

````python
def CreateUser(UserName, Email):
    UserName = UserName.strip().lower()
    Email = Email.strip().lower()

    if len(UserName) < 6 or "@" not in Email:
        print("Invalid user")
        return None

    print("Creating user:", UserName)

    return {
        "username": UserName,
        "email": Email,
    }


CreateUser("  Alex_21  ", " Alex@Example.COM ")
````

### Refactoring points

- Use `snake_case` for function and parameter names.
- Use names that describe intent.
- Keep one clear responsibility in each function.
- Separate normalization, validation, and user creation where useful.
- Prefer returning useful results over mixing reusable logic with output.

### Refactored




### Possible final structure

The refactored version can separate the work into focused functions:

```python
normalize_username(...)
normalize_email(...)
is_valid_user(...)
create_user(...)
```



````python
def normalize_username(user_name):
    return user_name.strip().lower()


def normalize_email(email):
    return email.strip().lower()


def is_valid_user(user_name, email):
    return len(user_name) >= 6 and "@" in email


def create_user(user_name, email):
    user_name = normalize_username(user_name)
    email = normalize_email(email)

    if not is_valid_user(user_name, email):
        return None

    return {
        "username": user_name,
        "email": email,
    }


user = create_user(
    "  Alex_21  ",
    " Alex@Example.COM ",
)

print(user)
````

## Default arguments

**Syntax**

```python
def function_name(required, optional=default_value):
    ...
```

A default value is used when the matching argument is omitted.

### Example 1: Minimum username length

````python
def is_valid_username(username, min_length=6):
    return (
        len(username) >= min_length
        and username.replace("_", "").isalnum()
    )


print(is_valid_username("alex_21"))
print(is_valid_username("sam", min_length=3))
````

### Example 2: Login attempt limit

````python
def can_retry(attempts_used, max_attempts=3):
    return attempts_used < max_attempts


print(can_retry(2))
print(can_retry(3))
print(can_retry(4, max_attempts=5))
````

# 2. Useful Python Tools

## `range()`

**Signature**

```python
range(stop, /)
range(start, stop, step=1, /)
```

`range` represents a sequence of integers. The stop value is excluded.

### Allow a fixed number of login attempts

`range(start, stop)` can represent the allowed attempt numbers. The stop value is excluded.

````python
max_attempts = 3

for attempt in range(1, max_attempts + 1):
    print(f"Login attempt {attempt} of {max_attempts}")
````

### Generate a sequence of user IDs

`range(start, stop)` starts at `start` and excludes `stop`.

````python
for user_id in range(1001, 1006):
    print(f"Creating test user {user_id}")
````

### Show 15-minute intervals

The third argument controls the step size. This range moves forward by 15 each time.

````python
for minute in range(0, 60, 15):
    print(f"{minute} minutes")
````

### Countdown

A negative step makes the sequence decrease.

````python
for number in range(5, 0, -1):
    print(number)

print("Start!")
````

### Empty range

A range is empty when its direction cannot reach the stop value.

````python
print(list(range(5, 5)))
print(list(range(5, 2)))
print(list(range(2, 5, -1)))
````

### Invalid step

A step of `0` raises `ValueError`.

````python
try:
    range(1, 5, 0)
except ValueError as error:
    print(type(error).__name__ + ":", error)
````

## `zip()`

**Signature**

```python
zip(*iterables, strict=False)
```

`zip` combines values from multiple iterables position by position.

### Example 1: Pair usernames and emails

````python
usernames = ["alex_21", "sam99", "maya_dev"]
emails = [
    "alex@example.com",
    "sam@example.com",
    "maya@example.com",
]

for username, email in zip(usernames, emails):
    print(username, email)
````

### Example 3: Different lengths

````python
usernames = ["alex_21", "sam99", "maya_dev"]
roles = ["admin", "user"]

print(list(zip(usernames, roles)))
````

### Example 4: `strict=True`

````python
usernames = ["alex_21", "sam99", "maya_dev"]
roles = ["admin", "user"]

try:
    print(list(zip(usernames, roles, strict=True)))
except ValueError as error:
    print(type(error).__name__ + ":", error)
````

## Iterators

Use `iter()` and `next()` to retrieve items one at a time.

````python
# iter() creates an iterator; next() retrieves one item at a time.
usernames = ["alex_21", "sam99", "maya_dev"]
user_iterator = iter(usernames)

print(next(user_iterator))
print(next(user_iterator))
print(next(user_iterator))
print(next(user_iterator, "No more users"))
````

## Generators

Use `yield` to produce values on demand rather than returning them all at once.

````python
# A generator produces values only when requested.
def login_attempt_numbers(max_attempts):
    for attempt in range(1, max_attempts + 1):
        print(f"Producing attempt {attempt}")
        yield attempt

attempts = login_attempt_numbers(3)
print("Generator created")
print(next(attempts))
print(next(attempts))
print(list(attempts))
````

## `*args`

**Syntax**

```python
def function_name(*args):
```

Extra positional arguments are collected into a tuple.

## `**kwargs`

**Syntax**

```python
def function_name(**kwargs):
    ...
```

Extra keyword arguments are collected into a dictionary.

### Example 1: Create a flexible user profile

````python
def create_profile(username, **details):
    profile = {
        "username": username,
        **details,
    }

    return profile


profile = create_profile(
    "alex_21",
    email="alex@example.com",
    role="admin",
    active=True,
)

print(profile)
````

## Lambda expressions

**Syntax**

```python
lambda arguments: expression
```

A lambda expression creates a small function containing one expression.

### Example 1: Sort users by score

````python
users = [
    {"username": "alex_21", "score": 82},
    {"username": "sam99", "score": 91},
    {"username": "maya_dev", "score": 88},
]

ordered = sorted(
    users,
    key=lambda user: user["score"],
    reverse=True,
)

print(ordered)
````

## Decorators

Use `@decorator_name` to add behavior around a function without changing its body.

````python
# A decorator adds logging without changing the original function.
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print("Call finished")
        return result

    return wrapper


@log_call
def get_user_status(username):
    return f"{username} is active"


print(get_user_status("alex_21"))
````

# 5. OOP Essentials

## Classes and objects

**Syntax**

```python
class ClassName:
    def __init__(self, ...):
        ...

    def method(self, ...):
        ...
```

A class defines a type. Each object can store its own state and use methods defined by the class.

### Example 1: User account

````python
class UserAccount:
    def __init__(self, username, email):
        self.username = username
        self.email = email
        self.active = True

    def display(self):
        return f"{self.username} <{self.email}>"

    def deactivate(self):
        self.active = False


user = UserAccount(
    "alex_21",
    "alex@example.com",
)

print(user.display())
print(user.active)

user.deactivate()

print(user.active)
````

### Example 2: Independent object state

````python
first = UserAccount("alex_21", "alex@example.com")
second = UserAccount("sam99", "sam@example.com")

first.deactivate()

print(first.username, first.active)
print(second.username, second.active)
````

### Example 3: Method with an argument

````python
class UserAccount:
    def __init__(self, username):
        self.username = username
        self.failed_attempts = 0

    def record_failed_login(self):
        self.failed_attempts += 1

    def is_locked(self, max_attempts=3):
        return self.failed_attempts >= max_attempts


user = UserAccount("sam99")

user.record_failed_login()
user.record_failed_login()
user.record_failed_login()

print(user.failed_attempts)
print(user.is_locked())
````

# 7. Exceptions in Python

## `try` and `except`

**Syntax**

```python
try:
    operation()
except SomeError:
    handler()
```

Exceptions represent runtime failures. Specific exceptions should be handled when the program knows how to respond.

### Example 1: Parse a user ID

````python
raw_user_id = "1024"

try:
    user_id = int(raw_user_id)
    print(user_id)
except ValueError:
    print("User ID must be numeric.")
````

### Example 2: Missing dictionary key

````python
user = {
    "username": "alex_21",
    "email": "alex@example.com",
}

try:
    print(user["role"])
except KeyError:
    print("Role is not available.")
````

### Example 3: Multiple exception types

````python
def get_user_age(record):
    try:
        return int(record["age"])
    except KeyError:
        return "Age field is missing."
    except ValueError:
        return "Age must be numeric."


print(get_user_age({"age": "24"}))
print(get_user_age({}))
print(get_user_age({"age": "unknown"}))
````

### Example 4: `else` and `finally`

````python
try:
    user_id = int("205")
except ValueError:
    print("Invalid user ID.")
else:
    print("User ID:", user_id)
finally:
    print("Validation finished.")
````

## `raise`

**Syntax**

```python
raise ExceptionType("message")
```

`raise` creates an exception explicitly when a requirement is violated.

### Example: Reject an invalid username

````python
def register_username(username):
    if len(username) < 6:
        raise ValueError(
            "Username must contain at least 6 characters."
        )

    return username


print(register_username("alex_21"))

try:
    register_username("sam")
except ValueError as error:
    print(type(error).__name__ + ":", error)
````
