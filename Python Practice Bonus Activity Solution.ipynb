Account Security Review

## Task

Review the user accounts below and create a simple security report.

An account needs review when it is **active** and either:

```python
failed_attempts >= 3
```

or MFA is disabled.

Each function below has one clear responsibility.

## Input

````python
users = [
    {
        "username": "alex_21",
        "email": "alex@example.com",
        "role": "user",
        "active": True,
        "failed_attempts": 1,
        "mfa_enabled": True,
    },
    {
        "username": "sam99",
        "email": "sam@example.com",
        "role": "admin",
        "active": True,
        "failed_attempts": 4,
        "mfa_enabled": False,
    },
    {
        "username": "maya_dev",
        "email": "maya@example.com",
        "role": "moderator",
        "active": True,
        "failed_attempts": 2,
        "mfa_enabled": True,
    },
    {
        "username": "dev_01",
        "email": "dev@example.com",
        "role": "user",
        "active": False,
        "failed_attempts": 5,
        "mfa_enabled": False,
    },
    {
        "username": "nina88",
        "email": "nina@example.com",
        "role": "user",
        "active": True,
        "failed_attempts": 4,
        "mfa_enabled": True,
    },
    {
        "username": "guest_7",
        "email": "guest@example.com",
        "role": "guest",
        "active": True,
        "failed_attempts": 0,
        "mfa_enabled": False,
    },
]
````

## 1. Check whether an account needs review

**Concepts:** functions, conditions, Boolean logic, dictionary access

````python
def needs_security_review(user):
    is_active = user["active"]

    has_many_failed_attempts = (
        user["failed_attempts"] >= 3
    )

    mfa_is_disabled = not user["mfa_enabled"]

    needs_review = (
        is_active
        and (
            has_many_failed_attempts
            or mfa_is_disabled
        )
    )

    return needs_review
````

````python
print(needs_security_review(users[0]))
print(needs_security_review(users[1]))

accounts_to_review = [
    user
    for user in users
    if needs_security_review(user)
]

accounts_to_review
````

## 2. Sort the accounts

**Concepts:** `sorted()`, `key=`, lambda, `reverse=True`, return values

````python
def sort_by_failed_attempts(users):
    sorted_users = sorted(
        users,
        key=lambda user: user["failed_attempts"],
        reverse=True,
    )

    return sorted_users
````

````python
sorted_accounts = sort_by_failed_attempts(
    accounts_to_review
)

sorted_accounts
````

## 3. Format the security report

**Concepts:** loops, `enumerate()`, f-strings, list building

````python
def format_security_report(users):
    report = []

    for number, user in enumerate(users, start=1):
        if user["mfa_enabled"]:
            mfa_status = "enabled"
        else:
            mfa_status = "disabled"

        line = (
            f'{number}. {user["username"]} | '
            f'{user["role"]} | '
            f'{user["failed_attempts"]} failed attempts | '
            f'MFA: {mfa_status}'
        )

        report.append(line)

    return report
````

````python
report = format_security_report(
    sorted_accounts
)

for line in report:
    print(line)
````

## 4. Check administrator MFA

**Concepts:** `all()`, generator expressions, filtering conditions

````python
def all_active_admins_have_mfa(users):
    result = all(
        user["mfa_enabled"]
        for user in users
        if user["active"] and user["role"] == "admin"
    )

    return result
````

````python
admin_mfa_status = all_active_admins_have_mfa(users)

print(
    "All active admins have MFA enabled:",
    admin_mfa_status,
)
````
